```python
from tkinter import *
from tkinter import ttk
```

```python
root = Tk()
root.geometry("300x300")

def calculate():
    try:
        b = float(w.get())/float(h.get())**2
        res.set(b)
        if b <= 16:
            intrp.set("Выраженный дефицит массы тела")
        elif b <= 18.5:
            intrp.set("Недостаточная (дефицит) масса тела")
        elif b <= 25:
            intrp.set("Норма")
        elif b <= 30:
            intrp.set("Избыточная масса тела (предожирение)")
        elif b <= 35:
            intrp.set("Ожирение 1 степени")
        elif b <= 40:
            intrp.set("Ожирение 2 степени")
        else:
            intrp.set("Ожирение 3 степени")
    except:
        pass

f = ttk.Frame(root)
f.grid(column=0, row=0, sticky=(N, W, E, S))
root.columnconfigure(0, weight=1)
root.rowconfigure(0, weight=1)
root.columnconfigure(2, weight=3)

w = StringVar()
weight_entry = Entry(f, textvariable=w)
weight_entry.grid(column=2, row=1, sticky=(W, E))

h = StringVar()
height_entry = Entry(f, textvariable=h)
height_entry.grid(column=2, row=2, sticky=(W, E))

Label(f, text="Weight").grid(column=1, row=1, sticky=W)
Label(f, text="Height").grid(column=1, row=2, sticky=W)


res = StringVar()
Label(f, textvariable=res).grid(column=1, row=3, sticky=(W, E))

intrp = StringVar()
Label(f, textvariable=intrp).grid(column=1, row=4, sticky=(W, E))

Button(f, text="Calculate", command=calculate).grid(column=1, row=5, columnspan=2, sticky=(W, E))

root.mainloop()
```
