```python
import pandas as pd
with open("imdb_top_250.csv", encoding="utf-8") as f:
    df = pd.read_csv(f)
genres = {}
for film, genre in zip(df["Title"], df["Genre"]):
    for i in genre.split(" | "):
        if i in genres:
            genres[i].append(film)
            continue
        genres[i] = [film]
```

```python
from tkinter import *
from tkinter import ttk
from random import choice
```

```python
root = Tk()
root.geometry("300x300")

def calculate():
    try:
        res.set(choice(genres[exp.get()]))
    except:
        pass

f = ttk.Frame(root, padding = "3 3 12 12")
f.grid(column=0, row=0,sticky=(N, W, E, S))
root.columnconfigure(0, weight=1)
root.rowconfigure(0, weight=1)

exp = StringVar()
exp_entry = OptionMenu(f, exp, *genres.keys())
exp_entry.grid(column=1, row=1, sticky=(W, E))

res = StringVar()
Label(f, textvariable=res).grid(column=1, row=2, sticky=(W, E))

Button(f, text="Calculate", command=calculate).grid(column=1, row=3, sticky=(W, E))

root.mainloop()
```
