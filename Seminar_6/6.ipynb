```python
from tkinter import *
from tkinter import ttk
from tkinter import filedialog
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python


root = Tk()
root.geometry("300x300")

def select():
    path = filedialog.askopenfile(defaultextension=".csv", filetypes=[("CSV", ".csv")])
    global df
    df = pd.read_csv(path)

def calculate():
    k, b = np.polyfit(df[df.columns[0]], df[df.columns[1]], 1)
    kxb.set(f"{k:.2g}*x+{b:.2g}")
    plt.cla()
    fig, ax = plt.subplots(figsize=(10, 10))
    ax.grid()
    ax.set_xlabel(df.columns[0])
    ax.set_ylabel(df.columns[1])
    ax.scatter(df[df.columns[0]], df[df.columns[1]], label="Значения")
    plt.tight_layout()
    ax.autoscale(enable=False, axis="x")
    ax.plot(ax.get_xlim(), np.array(ax.get_xlim())*k + b, label=f"$y = {k:.2g}\cdot x+{b:.2g}$")
    ax.legend()
    plt.savefig(filedialog.asksaveasfile(defaultextension=".png", filetypes=[("PNG", ".png")]).name, dpi=300)

f = ttk.Frame(root, padding="3 3 12 12")
f.grid(column=0, row=0, sticky=(N, W, E, S))
root.columnconfigure(0, weight=1)
root.rowconfigure(0, weight=1)
Button(f, text="Select", command=select).grid(column=1, row=1, sticky=(W, E))
Button(f, text="Calculate", command=calculate).grid(column=1, row=2, sticky=(W, E))
kxb = StringVar()
Label(f, textvariable=kxb).grid(column=1, row=3)

root.mainloop()
```
