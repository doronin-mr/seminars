```python
N, M = map(int, input().split())
dirr = 0
n = 0
m = 0
mat = [[0 for i in range(N)] for j in range(M)]
i = 1
while True:
    mat[m][n] = i
    i += 1
    match dirr:
        case 0:
            if n >= N - 1 or mat[m][n+1] != 0:
                dirr = (dirr + 1) % 4
                m += 1
            else:
                n += 1
        case 1:
            if m == M - 1 or mat[m+1][n] != 0:
                dirr = (dirr + 1) % 4
                n -= 1
            else:
                m += 1
        case 2:
            if mat[m][n-1] != 0:
                dirr = (dirr + 1) % 4
                m -= 1
            else:
                n -= 1
        case 3:
            if mat[m-1][n] != 0:
                dirr = (dirr + 1) % 4
                n += 1
            else:
                m -= 1
    if mat[m][n] != 0:
        break
print(*mat,sep="\n")
```

```python
import numpy as np
mat = np.array(mat)
rows = np.arange(M).reshape(-1, 1)
print(mat*rows)
```
