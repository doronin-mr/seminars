```python
def det(m):
    if len(m) == 1:
        return m[0][0]
    a = 0
    for n, i in enumerate(m[0]):
        minor = []
        for row in m[1:]:
            minor_row = []
            for n_col, cell in enumerate(row):
                if n_col != n:
                    minor_row.append(cell)
            minor.append(minor_row)
        a += (-1)**(n)*i*det(minor)
    return a

N, M = map(int,input().split())
A = []
b = []
for i in range(N):
    row = list(map(int, input().split()))
    A.append(row[:-1])
    b.append(row[-1])
delta = det(A)
A_n = []
for i in range(len(A[0])):
    an = [i.copy() for i in A]
    for j in range(len(an)):
        an[j][i] = b[j]
    A_n.append(det(an))
print(*list(i/delta for i in A_n))
```

```python
import numpy as np
N, M = map(int,input().split())
A = []
b = []
for i in range(N):
    row = list(map(int, input().split()))
    A.append(row[:-1])
    b.append(row[-1])
print(*np.linalg.solve(A, b))
```
