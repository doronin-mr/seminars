```python
def F(N, cache = {0: 0, 1: 1}):
    if N in cache:
        return cache[N]
    else:
        cache[N] = F(N-1, cache) + F(N-2, cache)
        return cache[N]

print(F(int(input()))
```

```python
def D(N):
    if N <= 1:
        return N
    a = 1
    b = 0
    for i in range(N-1):
        a, b = a + b, a
    return a

print(D(int(input())))
```
