```python
def prod(a):
    p = 1
    for i in a:
        p *= i
    return p

inp = open("input.txt", "r")
out = open("output.txt", "w")
nums = list(map(int, inp.readline().split()))
op = inp.readline().strip()
match op:
    case "+":
        out.write(str(sum(nums)))
    case "*":
        out.write(str(prod(nums)))
    case "-":
        out.write(str(nums[0] - sum(nums[1:])))
inp.close()
out.close()
```
