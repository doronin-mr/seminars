```python
def toN(N, c):
    res = ""
    while N>0:
        res = str(N % c) + res
        N //= c
    return res

def prod(a):
    p = 1
    for i in a:
        p *= i
    return p

inp = open("input.txt", "r")
out = open("output.txt", "w")
numN = inp.readline().split()
op = inp.readline().strip()
N = int(inp.readline())
nums = []
for i in numN:
    nums.append(int(i, N))
match op:
    case "+":
        out.write(toN(sum(nums), N))
    case "*":
        out.write(toN(prod(nums), N))
    case "-":
        out.write(toN(nums[0] - sum(nums[1:]), N))
inp.close()
out.close()
```
