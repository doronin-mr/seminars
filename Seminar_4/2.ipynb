```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
random_generator = np.random.default_rng(seed=67)

samples = [random_generator.normal(scale=5, size=k*100) for k in range(1, 11)]

fig, axs = plt.subplots(10, figsize=(10, 100))

for n, ax in enumerate(axs):
    ax.hist(samples[n], bins=np.arange(min(samples[n]), max(samples[n]) + 1, 1))
    ax.set_xlim(np.array([-1, 1])*max(np.absolute(ax.get_xlim())))
```
