```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("iris_data.csv")
fig, axs = plt.subplots(2, figsize=(10, 20))
label, perc = np.unique(df["Species"], return_counts=True)
axs[0].pie(perc, labels=label, autopct="%1.1f%%")
axs[1].pie([len(df["PetalLengthCm"][df["PetalLengthCm"]<=1.2]), len(df["PetalLengthCm"][(1.5>=df["PetalLengthCm"]) & (df["PetalLengthCm"]>1.2)]), len(df["PetalLengthCm"][df["PetalLengthCm"]>1.5])], \
          labels = ["<=1.2", "(1.2, 1.5]", ">1.5"], autopct="%1.1f%%")
plt.show()
```
