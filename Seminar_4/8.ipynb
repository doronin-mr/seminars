```python
import numpy as np
```

```python
a = list(np.random.randint(1, 100, 100))
b = list(np.random.randint(1, 100, 100))
```

```python
a_s = set(a)
b_s = set(b)
uniq = set()
for i in a_s:
    if a.count(i) == 1:
        print(i, end=" ")
        uniq.add(i)
print()
for i in b_s:
    if b.count(i) == 1:
        print(i, end=" ")
        if i in uniq:
            uniq.discard(i)
        else:
            uniq.add(i)
        continue
    uniq.discard(i)
print()
print(*uniq)
print(*(a_s & b_s))
```
