```python
mirrors      = ["A", "H", "I", "M", "O", "T", "U", "V", "W", "X", "Y", "1", "8"]
have_mirrors = ["E", "J", "S", "Z", "3", "L", "2", "5"]
mirror_n     = {"E": "3", "J": "L", "S": "2", "Z": "5", "3": "E", "L": "J", "2": "S", "5": "Z"}

s = input()

message = {(0, 0): f"{s} is not a palindrome.", (1, 0): f"{s} is a regular palindrome.", \
           (0, 1): f"{s} is a mirrored string.", (1, 1): f"{s} is a mirrored palindrome."}

mirror_f = 1
for i in range(len(s)):
    if s[i] in mirrors:
        continue
    if s[i] in have_mirrors and mirror_n[s[len(s)-1-i]] == s[i]:
        continue
    mirror_f = 0
    break
palindrome = int(s == s[::-1])
print(message[(palindrome, mirror_f)])
```
