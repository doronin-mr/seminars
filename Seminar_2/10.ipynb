```python
with open("input.txt", encoding="utf-8") as f:
    s = f.read()
gl  = "аоуеэяиюАОУЕЭЯИЮ"
sgl = list(set("йцуцкенгшщзхъфыфвапарпорлоджэжячясмсимтиьбюЙЦУКЕНГШЩШЗХЪФЫВАПРОЛДЖЭЯЧСМИТЬБЮ").difference(gl))
res = ""
for n, i in enumerate(s):
    if i in gl and s[n-1] in sgl:
        res += f"{i}с{i}"
    else:
        res += i
print(res)
```
