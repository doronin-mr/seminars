```python
s1 = input()
s2 = input()
a = [[0 for i in range(len(s1)+1)] for j in range(len(s2)+1)]
for y, row in enumerate(a):
    for x, elem in enumerate(row):
        if x == 0:
            if y == 0:
                continue
            a[y][x] = a[y-1][x] + 1
            continue
        if y == 0:
            a[y][x] = a[y][x-1] + 1
            continue
        a[y][x] = min(a[y][x-1] + 1, a[y-1][x] + 1, a[y-1][x-1] if s1[x-1]==s2[y-1] else a[y-1][x-1] + 1)
print(a[-1][-1])
```
