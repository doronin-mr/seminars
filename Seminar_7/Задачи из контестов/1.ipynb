```python
N = int(input())
a = [float("inf") for i in range(N)]
a[0] = 0
for n in range(len(a)-1):
    a[n+1] = min(a[n],\
            a[(n+2)//2 - 1] if (n+2)% 2 == 0 else float("inf"),\
            a[(n+2)//3-1] if (n+2)%3==0 else float("inf")) + 1
print(a[-1])
```

```python
N = int(input())
res = 0
while N > 1:
    while N % 3 == 0:
        N //= 3
        res += 1
    while N % 2 == 0:
        N //= 2
        res += 1
    if N > 1:
        N -= 1
        res += 1
print(res)
```
