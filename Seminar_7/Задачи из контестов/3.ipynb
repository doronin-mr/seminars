```python
N, K = map(int, input().split())
sec_map = []
for i in range(N):
    sec_map.append(list(map(int, input().split())))
```

```python
m = 0
for row, r in enumerate(sec_map):
    for col, cell in enumerate(r):
        if cell == 1:
            if row == 0 or col == 0:
                m = 1 if 1 > m else m
                continue
            sec_map[row][col] = min(sec_map[row - 1][col], sec_map[row - 1][col - 1], sec_map[row][col - 1]) + 1
            m = sec_map[row][col] if sec_map[row][col] > m else m
print(m)
```
